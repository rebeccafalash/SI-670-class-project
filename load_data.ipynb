```python
from pathlib import Path
import pandas as pd

csv_dir = Path("csv")
dfs = {}
for f in sorted(csv_dir.glob("*.csv")):
    dfs[f.stem] = pd.read_csv(f, low_memory=False)
    print(f"{f.name}: {dfs[f.stem].shape}")
```

```python
# Optional: combine all years into one DataFrame
all_data = pd.concat(
    [df.assign(source_file=name) for name, df in dfs.items()],
    ignore_index=True,
)
all_data.shape
```
